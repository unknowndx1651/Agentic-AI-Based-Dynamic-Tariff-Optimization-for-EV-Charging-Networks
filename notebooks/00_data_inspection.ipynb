In this notebook, the datasets are loaded and inspected.  
For the ACN Dataset, important keys are observed, and for the Urban EV Dataset, the contents and shape of each data set are observed.

```python
# ACN Dataset Inspection
import json
import pandas as pd

with open('../data/raw/acndata_sessions.json', 'r') as f:
    acn = json.load(f)

# See the top-level keys
print("Top-level keys:", list(acn.keys()))

# See how deep it goes
for key in acn.keys():
    val = acn[key]
    print(f"\nKey: '{key}' → Type: {type(val)}")
    if isinstance(val, list):
        print(f"  Length: {len(val)}")
        print(f"  First item keys: {list(val[0].keys()) if val else 'empty'}")
    elif isinstance(val, dict):
        print(f"  Sub-keys: {list(val.keys())[:5]}")
        

```
